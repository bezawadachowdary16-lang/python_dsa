```python
def factorial(n):
  if n==0 or n==1:
    return 1
  else:
    return n*factorial(n-1)
print(factorial(5))
```

```python
#Queue
from collections import deque
queue=deque()
queue.append(1)
queue.append(2)
queue.popleft()
print(queue)
```
