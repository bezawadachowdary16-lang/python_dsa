```python
arr=[2,4,1,5,3,7,2]
sum=0
for i in range(len(arr)):
  sum+=arr[i]
print(sum)
```

```python
def build_prefix(arr):
  prefix=[arr[0]]
  for i in range(1,len(arr)):
    prefix.append(prefix[i-1]+arr[i])
  return prefix
print(build_prefix(arr))
```

```python
#sum of range 2 to 5 th index
def range_sum(prefix,left,right):
  if left==0:
    return prefix[right]
  return prefix[right]-prefix[left-1]
prefix=build_prefix(arr)
print(range_sum(prefix,2,5))
#prefix=[1,3,6,10] prefix[3]-prefix[1-1]==0
```

```python
# equillibrium index
arr1=[-7,-1,5,2,-4,3,0]
print(sum(arr1))
def equillibrium_index(n):
  left_sum=0
  total= sum(n)
  for i in range(len(n)):
    right_sum=total-n[i]-left_sum
    if left_sum==right_sum:
      return i
    left_sum+=n[i]
  return -1
arr1=[-7,-1,5,2,-4,3,0]
print(equillibrium_index(arr1))
#
```
