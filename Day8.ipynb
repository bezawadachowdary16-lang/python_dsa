```python
#Sort the array
arr=list(map(int,input().split()))
n=len(arr)
for i in range(n):
  for j in range(0,n-i-1):
    if arr[j]>arr[j+1]:
      arr[j],arr[j+1]=arr[j+1],arr[j]
print(arr)
```

```python
arr=[3,4,2,1,5,8,7]
def merge_sort(arr):
  if len(arr)>1:
    mid=len(arr)//2
    L=arr[:mid]
    R=arr[mid:]
    merge_sort(L)
    merge_sort(R)
    lp=0
    rp=0
    fp=0
    while lp<len(L) and rp<len(R):
      if L[lp]<R[rp]:
        arr[fp]=L[lp]
        lp+=1
      else:
        arr[fp]=R[rp]
        rp+=1
      fp+=1
    while lp<len(L):
      arr[fp]=L[lp]
      lp+=1
      fp+=1
    while rp<len(R):
      arr[fp]=R[rp]
      rp+=1
      fp+=1
merge_sort(arr)
print(arr)
```

```python
#Quick sort
import random
arr=[3,4,2,1,5,8,7]
def quick_sort(arr):
  if len(arr)<=1:
    return arr
  pivot=random.choice(arr)
  left=[i for i in arr if i<pivot]
  right=[i for i in arr if i>pivot]
  return quick_sort(left)+[pivot]+quick_sort(right)
print(quick_sort(arr))



```
