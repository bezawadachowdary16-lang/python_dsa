```python
#find the non repeating element
arr=list(map(int,input().split()))
freq_dict={}
for i in arr:
  if i in freq_dict:
    freq_dict[i]+=1
  else:
    freq_dict[i]=1
for j in arr:
  if freq_dict[j]==1:
    print(j)
    break
```

```python
# find duplicates and print
n11=list(map(int,input().split()))
d2={}
for i in n11:
  if i in d2:
    d2[i]+=1
  else:
    d2[i]=1
for j in d2:
  if d2[j]>1:
    print(j)
```

```python
#find duplicates using set
n22=list(map(int,input().split()))
seen=set()
res=set()

for i in n22:
  if i not in seen:
    seen.add(i)
  else:
    res.add(i)
print(res)
```
