```python
# get input from the user as string
# remove duplicates characters from the string
x=input()
seen=set()
result=""
for i in x:
  if i in seen:
    continue
  seen.add(i)
  result+=i
print(result)
```

```python
# get input from user reverse it
# use for loop no slicing [::-1] dont use
x=input()
reverse=""
for i in x:
    reverse=i+reverse
print(reverse)
```

```python
# Recursion
# factorial
def factorial(n):
  if n <= 1:
    return 1
  return n*factorial(n-1)
factorial(5)
```
