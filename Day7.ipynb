```python
`#finding elements in a array
arr=list(map(int,input().split()))
target=int(input())
found=False
for i in range(len(arr)):
  if target==arr[i]:
    print("Target found at index",i)
    found=True
    break
if not found:
  print("target not found")
```

```python
#use Binary search
a1=list(map(int,input().split()))
target=int(input())
a1.sort()
high=len(a1)-1
low=0
while low<=high:
  mid=(high+low)//2
  if target==a1[mid]:
    print("target found at index",mid)
    break
  elif target>a1[mid]:
    low=mid+1
  else:
    high=mid-1
```

```python
#calculate square root of 50 using binary search
target=int(input())
precision=0.00001
low=0.0
high=float(target)
while(high-low>precision):
  mid=(high+low)/2
  if mid*mid==target:
    print(mid)
    break
  elif mid*mid>target:
    high=mid
  else:
    low=mid
print(mid)
```
